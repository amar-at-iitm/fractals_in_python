$$
I=[x_0,x_N]\subset \mathbb{R}, \qquad x_0<x_1<\cdots <x_N .
$$

Let

$$
f\in C(I)
$$

be the original function, and let

$$
g\in C(I)
$$

be a base function satisfying the endpoint interpolation conditions

$$
g(x_0)=f(x_0), \qquad g(x_N)=f(x_N).
$$

Let

$$
\Delta=\{x_0,x_1,\ldots,x_N\}
$$

be a partition of $I$. For each $i=1,2,\ldots,N$, define an affine map

$$
L_i:I\to I_i=[x_{i-1},x_i]
$$

by

$$
L_i(x) = a_i x+b_i,
$$

where

$$
a_i=\frac{x_i-x_{i-1}}{x_N-x_0},
$$

and

$$
b_i=\frac{x_Nx_{i-1}-x_0x_i}{x_N-x_0}.
$$

Thus,

$$
L_i(x_0)=x_{i-1}, \qquad L_i(x_N)=x_i.
$$

The inverse map is

$$
L_i^{-1}(x)=\frac{x-b_i}{a_i}, \qquad x\in [x_{i-1},x_i].
$$

Equivalently,

$$
L_i^{-1}(x) = x_0 + \frac{(x-x_{i-1})(x_N-x_0)}{x_i-x_{i-1}}.
$$

Let

$$
\alpha=(\alpha_1,\alpha_2,\ldots,\alpha_N)
$$

be a scale vector satisfying

$$
|\alpha_i|<1, \qquad i=1,2,\ldots,N.
$$

The $\alpha$-fractal function associated with $f$, the base function $g$, the partition $\Delta$, and the scale vector $\alpha$ is the unique continuous function

$$
f^\alpha:I\to \mathbb{R}
$$

satisfying the self-referential equation

$$
f^\alpha(x)=f(x)+\alpha_i\left[f^\alpha\left(L_i^{-1}(x)\right)-g\left(L_i^{-1}(x)\right)\right], \qquad x\in [x_{i-1},x_i],
$$

for every

$$
i=1,2,\ldots,N.
$$


Equivalently, the self-referential equation may be written in the form


$$
f^\alpha(L_i(x)) = f(L_i(x)) + \alpha_i \left[f^\alpha(x)-g(x)\right], \qquad x\in I,\quad i=1,2,\ldots,N.
$$


In particular, for the nodal points


$$
x_j\in \Delta,\qquad j=0,1,2,\ldots,N,
$$


we get


$$
f^\alpha(L_i(x_j)) = f(L_i(x_j)) + \alpha_i \left[f^\alpha(x_j)-g(x_j)\right], \qquad i=1,2,\ldots,N,\quad j=0,1,2,\ldots,N.
$$


Thus, at the first stage of iteration, we obtain the set of points


$$
\left\{\left(L_i(x_j),\,f^\alpha(L_i(x_j))\right): i=1,2,\ldots,N,\; j=0,1,2,\ldots,N\right\},
$$


which consists of at most


$$
N(N+1)
$$


points.

Applying the same functional equation once more, for


$$
m=1,2,\ldots,N,
$$


we have


$$
f^\alpha(L_m(L_i(x_j)))
=
f(L_m(L_i(x_j)))
+
\alpha_m\left[f^\alpha(L_i(x_j))-g(L_i(x_j))\right].
$$


Substituting the previous relation, we get


$$
\begin{aligned}
f^\alpha(L_m(L_i(x_j)))
&=
f(L_m(L_i(x_j)))
+
\alpha_m\left[
f(L_i(x_j))
+
\alpha_i\left(f^\alpha(x_j)-g(x_j)\right)
-
g(L_i(x_j))
\right] \\
&=
f(L_m(L_i(x_j)))
+
\alpha_m\left[f(L_i(x_j))-g(L_i(x_j))\right]
+
\alpha_m\alpha_i\left[f^\alpha(x_j)-g(x_j)\right].
\end{aligned}
$$


Hence, after the second stage of iteration, we obtain the set


$$
\left\{\left(L_m(L_i(x_j)),\,f^\alpha(L_m(L_i(x_j)))\right): m,i=1,2,\ldots,N,\; j=0,1,2,\ldots,N\right\},
$$


which consists of at most


$$
N^2(N+1)
$$


points.

Proceeding inductively, let


$$
\sigma=(i_1,i_2,\ldots,i_k)\in \{1,2,\ldots,N\}^k
$$


be a multi-index, and define


$$
L_\sigma=L_{i_1}\circ L_{i_2}\circ \cdots \circ L_{i_k}.
$$


Then, by repeated application of the self-referential equation, we obtain


$$
\begin{aligned}
f^\alpha(L_\sigma(x))
&=
f(L_\sigma(x))
+
\alpha_{i_1}\left[f(L_{i_2}\circ\cdots\circ L_{i_k}(x)) - g(L_{i_2}\circ\cdots\circ L_{i_k}(x))\right] \\
&\quad
+
\alpha_{i_1}\alpha_{i_2}\left[f(L_{i_3}\circ\cdots\circ L_{i_k}(x)) - g(L_{i_3}\circ\cdots\circ L_{i_k}(x))\right] \\
&\quad + \cdots +
\left(\prod_{r=1}^{k-1}\alpha_{i_r}\right)\left[f(L_{i_k}(x))-g(L_{i_k}(x))\right] \\
&\quad +
\left(\prod_{r=1}^{k}\alpha_{i_r}\right)\left[f^\alpha(x)-g(x)\right].
\end{aligned}
$$


In particular, for


$$
x=x_j,\qquad j=0,1,2,\ldots,N,
$$


we get


$$
\begin{aligned}
f^\alpha(L_\sigma(x_j))
&=
f(L_\sigma(x_j))
+
\alpha_{i_1}\left[f(L_{i_2}\circ\cdots\circ L_{i_k}(x_j)) - g(L_{i_2}\circ\cdots\circ L_{i_k}(x_j))\right] \\
&\quad
+
\alpha_{i_1}\alpha_{i_2}\left[f(L_{i_3}\circ\cdots\circ L_{i_k}(x_j)) - g(L_{i_3}\circ\cdots\circ L_{i_k}(x_j))\right] \\
&\quad + \cdots +
\left(\prod_{r=1}^{k-1}\alpha_{i_r}\right)\left[f(L_{i_k}(x_j))-g(L_{i_k}(x_j))\right] \\
&\quad +
\left(\prod_{r=1}^{k}\alpha_{i_r}\right)\left[f^\alpha(x_j)-g(x_j)\right].
\end{aligned}
$$


Therefore, after the kth stage of iteration, we obtain the set of points


$$
\left\{\left(L_\sigma(x_j),\,f^\alpha(L_\sigma(x_j))\right): \sigma\in \{1,2,\ldots,N\}^k,\; j=0,1,2,\ldots,N\right\},
$$


which consists of at most


$$
N^k(N+1)
$$


points.

Accordingly, if the iteration is continued up to level


$$
K,
$$


then the graph of the $\alpha$-fractal function may be approximated by the finite collection of points generated recursively from the relation


$$
f^\alpha(L_i(x)) = f(L_i(x)) + \alpha_i\left[f^\alpha(x)-g(x)\right], \qquad i=1,2,\ldots,N.
$$
